```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
```

## 0️⃣ Background: AI, ML, DL and Data Science

- **Artificial Intelligence (AI):** the broad field of building systems that perform tasks normally requiring human intelligence (reasoning, perception, decision-making).
- **Machine Learning (ML):** a subset of AI where a system *learns patterns from data* instead of being given explicit rules. Logistic Regression, used in this assignment, is a classic supervised ML algorithm.
- **Deep Learning (DL):** a subset of ML that uses multi-layer neural networks to automatically learn feature representations from large amounts of data. DL is **not** used here — Logistic Regression is a simpler, linear, and highly interpretable model, which is preferable for a small, structured medical dataset like this one.
- **Data Science:** the broader practice of extracting knowledge from data using statistics, ML, visualization, and domain knowledge. This notebook follows the standard Data Science / ML pipeline: data collection → cleaning → feature design → algorithm selection → loss function selection → training → evaluation.

In this notebook, we apply this pipeline to a **medical binary classification problem**: predicting whether a patient has heart disease using clinical attributes such as cholesterol, blood pressure, chest pain type, etc.

# Task 1

## 1️⃣ Data Retrieval and Collection

```python
data = pd.read_csv("Heart_Disease_Prediction.csv")
data.head()
```

```python
data.describe()
```

**Dataset description:** This is the Heart Disease Prediction dataset (UCI Cleveland Heart Disease data, Kaggle version). It contains **270 patient records** and **14 columns**: 13 clinical/demographic features (Age, Sex, Chest pain type, BP, Cholesterol, FBS over 120, EKG results, Max HR, Exercise angina, ST depression, Slope of ST, Number of vessels fluro, Thallium) and 1 target column, `Heart Disease`, which is `Presence` or `Absence`. Each row represents one patient. The `describe()` output shows the numeric features are on very different scales (e.g. Cholesterol ranges up to ~560 while Sex is 0/1), which is important context for Task 2, where feature scaling will be needed.

## 2️⃣ Data Cleaning

```python
data.isna().sum()
```

There are **no missing values** in any column, so no imputation or row-dropping is required.

```python
data.dtypes
```

All feature columns are numeric (`int64`/`float64`); only the target `Heart Disease` is currently a categorical `object` (text), which must be converted to a binary 0/1 integer before modelling (done below).

```python
data["Heart Disease"].unique()
```

```python
# Check for invalid (non-physiological) Cholesterol values, e.g. 0 or negative readings
print("Minimum Cholesterol value:", data["Cholesterol"].min())
print("Number of rows with Cholesterol <= 0:", (data["Cholesterol"] <= 0).sum())
```

A cholesterol reading of 0 mg/dl is not physiologically possible for a living patient and would indicate a sensor/recording error. In this dataset there are no such invalid (≤ 0) values, so no rows need to be removed or imputed on that basis.

```python
# Mapping the object 'Presence' and 'Absence' of the binary target variable to int 1 and 0 respectively

data["Heart Disease"] = data["Heart Disease"].map({"Presence":1, "Absence":0})
```

The target variable `Heart Disease` is now encoded as a binary integer: **1 = Presence** of heart disease, **0 = Absence**. This satisfies the requirement that the target be binary and numeric for Logistic Regression.

## 3️⃣ Feature Design

```python
X_s = data[["Cholesterol"]]
y_s = data["Heart Disease"]   # Series, not DataFrame -> avoids sklearn DataConversionWarning
```

**Why Cholesterol?** High serum cholesterol (especially LDL, "bad" cholesterol) is a well-established clinical risk factor for atherosclerosis — the buildup of fatty plaques in the arteries — which restricts blood flow to the heart and is a direct precursor to coronary heart disease. Because of this established medical link, Cholesterol is a reasonable single predictor to test in Task 1, even though on its own it cannot capture the full picture of a patient's cardiovascular risk.

## 4️⃣ Algorithm Selection

```python
model_s = LogisticRegression()
```

**Why Logistic Regression?** `Heart Disease` is a *binary* outcome (0/1). Logistic Regression is designed exactly for this: instead of predicting a continuous value like Linear Regression, it passes a linear combination of the input(s) through the **sigmoid function**, squashing the output into the range (0, 1). This output is interpreted as the *probability* that the patient belongs to class 1 (Presence of heart disease). A decision threshold (default 0.5) is then used to convert that probability into a class prediction. This makes Logistic Regression simple, fast, and — importantly for a medical use case — interpretable, since each coefficient shows the direction and (after exponentiating) the odds-ratio effect of a feature on disease risk.

## 5️⃣ Loss Function Selection

```python
# By default it is log loss
```

**Binary Cross-Entropy (Log Loss):** scikit-learn's `LogisticRegression` is trained by minimizing log loss by default. For a single example with true label *y* ∈ {0,1} and predicted probability *p̂*, the log loss is:

L = −[ y·log(p̂) + (1−y)·log(1−p̂) ]

- If the true label is 1 and the model predicts a probability close to 1, the loss is close to 0 (good).
- If the true label is 1 but the model predicts a probability close to 0, the loss becomes very large (heavily penalized).
- The total loss minimized during training is the **average** of this expression over all training examples.

Log loss is preferred over something like Mean Squared Error for classification because it is convex for logistic regression (guaranteeing a well-behaved optimization) and because it directly penalizes confident-but-wrong probability predictions much more harshly than mildly-wrong ones.

## 6️⃣ Model Learning (Training)

```python
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_s, y_s, test_size=0.2, random_state=42)
```

```python
model_s.fit(X_train_s, y_train_s)
```

**How parameters are learned:** The data is split 80% train / 20% test (`test_size=0.2`) so the model is evaluated on unseen patients. During `fit()`, scikit-learn uses an iterative optimizer (LBFGS by default) to search for the intercept and coefficient values that **minimize the average log loss** computed on the training data. Starting from initial values, the optimizer repeatedly adjusts the parameters in the direction that reduces the loss (gradient-based updates) until the loss stops improving (convergence) or a maximum number of iterations is reached.

## 7️⃣ Model Evaluation

```python
y_pred_s = model_s.predict(X_test_s)
accuracy_s = accuracy_score(y_test_s, y_pred_s)
precision_s = precision_score(y_test_s, y_pred_s)
recall_s = recall_score(y_test_s, y_pred_s)
f1_s = f1_score(y_test_s, y_pred_s)
cm_s = confusion_matrix(y_test_s, y_pred_s)
print(f"""Accuracy : {accuracy_s*100:.2f} %\n
Precision : {precision_s:.6f} \n
Recall : {recall_s:.6f} \n
F1 Score : {f1_s:.6f}""")
```

```python
print(f"Intercept: {model_s.intercept_[0]:.4f}")
print(f"Coefficient (Cholesterol): {model_s.coef_[0][0]:.4f}")
```

**Interpretation of Task 1 results:** Using Cholesterol alone gives the metrics printed above. A positive coefficient on Cholesterol means higher cholesterol is associated with a higher predicted probability of heart disease, consistent with medical understanding. However, accuracy/precision/recall on a *single feature* are limited — Cholesterol overlaps heavily between patients with and without heart disease, so this model only captures part of the picture. The confusion matrix below shows how many patients were correctly/incorrectly classified into each class; in particular, watch the **false negative** count (patients with heart disease predicted as healthy), which is the most clinically risky type of error.

```python
disp = ConfusionMatrixDisplay(confusion_matrix= cm_s)
disp.plot(cmap=plt.cm.Blues)

plt.title("Confusion Matrix with a Single Feature")
plt.show()
```

The confusion matrix confirms that a single feature (Cholesterol) cannot perfectly separate the two classes — there is visible overlap, leading to a non-trivial number of misclassifications in both directions.

## Sigmoid Curve

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
theta = np.array([model_s.intercept_[0], model_s.coef_[0][0]])


plt.figure(figsize=(14,7))

plt.scatter(X_train_s, y_train_s, color="red", marker="x", label="Data")

x_values = np.linspace(X_s["Cholesterol"].min()-1, X_s["Cholesterol"].max()+1, 200)[:, None]
plt.plot(x_values, sigmoid(np.insert(x_values,0,1,axis=1).dot(theta[:,None])), color="blue", label="Sigmoid Curve")
plt.xlabel("Cholesterol")
plt.ylabel("Heart Disease")
plt.title("Logistic Regression Curve")

plt.tight_layout()
plt.show()
```

The plotted sigmoid (S-shaped) curve shows the predicted probability of heart disease as Cholesterol increases. The red markers (actual 0/1 labels) are scattered across the whole range of Cholesterol, which visually confirms that Cholesterol alone only weakly separates the two classes — the curve rises gradually rather than sharply, indicating low confidence single-feature predictions.

# Task 2

## 1️⃣ Data Retrieval and Collection

```python
data = pd.read_csv("Heart_Disease_Prediction.csv")
display(data.head())
display(data.describe())
```

Same dataset as Task 1 (270 rows, 14 columns), reloaded here for a self-contained Task 2 section.

## 2️⃣ Data Cleaning

```python
data["Heart Disease"].unique()
```

```python
# Mapping the object 'Presence' and 'Absence' of the binary target variable to int 1 and 0 respectively

data["Heart Disease"] = data["Heart Disease"].map({"Presence":1, "Absence":0})
```

As in Task 1, the target is mapped to 1 (Presence) / 0 (Absence). There are no missing values and no invalid Cholesterol/BP readings in this dataset (verified in Task 1), so the data is otherwise already clean.

## 3️⃣ Feature Design

```python
X_a = data.drop(["Heart Disease"], axis=1)
y_a = data["Heart Disease"]
```

**Using all features:** All 13 remaining columns are used as predictors this time. In this particular dataset, the categorical attributes (Sex, Chest pain type, FBS over 120, EKG results, Exercise angina, Slope of ST, Thallium) are **already encoded as numeric codes** in the source CSV, so no additional one-hot encoding is required for scikit-learn to accept them. (In a dataset where categories were given as text labels, `pd.get_dummies()` or `OneHotEncoder` would be needed first.)

**Why might more features help?** Heart disease risk is influenced by many interacting factors (age, blood pressure, chest pain type, exercise tolerance, etc.), not cholesterol alone. Giving the model more relevant variables lets it find combinations of risk factors that are more predictive than any single one, which should improve discriminative power — as long as the extra features are not just noise.

## 4️⃣ Feature Engineering

```python
X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(X_a, y_a, test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train_a = scaler.fit_transform(X_train_a)
X_test_a = scaler.transform(X_test_a)
```

**Why scale the features?** The features are on very different numeric scales (e.g. Age ~29–77 vs Cholesterol ~120–560 vs Sex 0/1). Logistic Regression's optimizer and its regularization term are sensitive to feature scale — features with larger raw magnitudes can dominate the loss surface and slow or distort convergence. `StandardScaler` rescales every feature to mean 0 and standard deviation 1, so all features contribute on a comparable footing. Note that the scaler is **fit only on the training data** and then applied to the test data, to avoid leaking test-set information into training.

## 5️⃣ Algorithm Selection

```python
model_a = LogisticRegression()
```

As in Task 1, Logistic Regression is used so the results are directly comparable; only the input feature set changes.

##  6️⃣ Loss Function Selection

```python
# By default it is log loss
```

Same loss function as Task 1 — Binary Cross-Entropy / log loss — so any difference in performance between Task 1 and Task 2 comes from the **feature set**, not the optimization objective.

## 7️⃣ Model Learning (Training)

```python
model_a.fit(X_train_a, y_train_a)
```

The model now learns one coefficient per input feature (13 coefficients + 1 intercept) by minimizing average log loss on the scaled training data, using the same 80/20 train-test split proportions as Task 1.

## 8️⃣ Model Evaluation

```python
y_pred_a = model_a.predict(X_test_a)
accuracy_a = accuracy_score(y_test_a, y_pred_a)
precision_a = precision_score(y_test_a, y_pred_a)
recall_a = recall_score(y_test_a, y_pred_a)
f1_a = f1_score(y_test_a, y_pred_a)

cm_a = confusion_matrix(y_test_a, y_pred_a)

print(f"""Accuracy : {accuracy_a*100:.2f} %\n
Precision : {precision_a:.6f} \n
Recall : {recall_a:.6f} \n
F1 Score : {f1_a:.6f}""")
```

**Interpretation of Task 2 results:** Using all 13 features typically improves accuracy, precision, recall, and F1-score compared to the single-feature model, because the model can now combine multiple weaker signals (age, chest pain type, max heart rate, ST depression, etc.) into a stronger overall prediction. Compare the printed metrics above with the Task 1 metrics — the multi-feature model is expected to better balance correctly identifying patients **with** heart disease (recall) without raising too many false alarms (precision).

```python
disp = ConfusionMatrixDisplay(confusion_matrix=cm_a)
disp.plot(cmap=plt.cm.Blues)

plt.title("Confusion Matrix with Multiple Feature")
plt.show()
```

Compared to the Task 1 confusion matrix, the diagonal (correct predictions) should be denser and the off-diagonal (errors) sparser, reflecting the improved separability gained from using all 13 features instead of just Cholesterol.

# 📊 Model Comparison: Task 1 vs Task 2

The cell below prints a side-by-side comparison table of the two models using the metrics already computed above.

```python
import pandas as pd

comparison = pd.DataFrame({
    "Single Feature (Cholesterol)": [accuracy_s, precision_s, recall_s, f1_s],
    "Multiple Features (All)": [accuracy_a, precision_a, recall_a, f1_a]
}, index=["Accuracy", "Precision", "Recall", "F1-score"])

comparison
```

**Which model performs better, and why?**
The multi-feature model (Task 2) generally outperforms the single-feature model (Task 1) across accuracy, precision, recall, and F1-score (see the table above — re-run the notebook and compare the actual printed numbers). This makes sense because heart disease has many contributing risk factors; Cholesterol alone only partially explains the outcome, while combining Age, Chest pain type, Max HR, ST depression, Thallium, etc. lets the model draw on a much richer, more medically realistic picture of each patient.

**How does adding more features affect accuracy and recall?**
Accuracy improves because the decision boundary can now use multiple dimensions to separate the two classes rather than a single noisy threshold on Cholesterol. Recall (the fraction of actual heart-disease patients correctly identified) typically improves the most, since features like Chest pain type, ST depression, Number of vessels fluro, and Thallium are clinically strong indicators of heart disease — arguably stronger than Cholesterol alone — and including them helps the model catch more true positive cases. This matters a lot in a medical context, where a missed diagnosis (false negative) is usually far more costly than a false alarm.

**Trade-offs between interpretability and performance:**
- The single-feature model is extremely **interpretable**: there is exactly one coefficient, and its sign/magnitude directly tells you "higher cholesterol → higher predicted risk," which is easy to explain to a non-technical audience (e.g. a doctor or patient). Its downside is limited predictive power.
- The multi-feature model is more **predictive** but harder to interpret at a glance — with 13 coefficients (on standardized features), understanding *why* the model made a specific prediction requires looking at all coefficients together, and multicollinearity between features (e.g. Age and Max HR) can make individual coefficients less intuitive.
- In practice, the right choice depends on the use case: a quick screening heuristic might favor simplicity, while a real diagnostic-support tool would favor the higher-performing multi-feature model, possibly paired with explainability tools (e.g. feature importance, SHAP) to recover some interpretability.
